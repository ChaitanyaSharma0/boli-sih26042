```python
!pip install transformers torch scipy -q

from transformers import VitsModel, AutoTokenizer
import torch, scipy

model = VitsModel.from_pretrained("facebook/mms-tts-hoc")
tokenizer = AutoTokenizer.from_pretrained("facebook/mms-tts-hoc")

text = "hon ko sakam re ol tan a"   # placeholder Ho text

inputs = tokenizer(text, return_tensors="pt")
with torch.no_grad():
    output = model(**inputs).waveform

scipy.io.wavfile.write("ho_output.wav", rate=model.config.sampling_rate,
                       data=output.float().numpy().T)

from IPython.display import Audio
Audio("ho_output.wav", rate=model.config.sampling_rate)
```

```python
from transformers import VitsModel, AutoTokenizer
import torch, scipy

samples = {
    "hoc": ("Ho",      "ଦା ଆଲେ ଜୀଉ ତାନା"),
    "unr": ("Mundari", "ଦା ଆଲେ ଜିଉ ତାନା"),
    "kru": ("Kurukh",  "अम्म हमक जीवन रअदा"),
    "sck": ("Sadri",   "पानी हमन के जीवन हे"),
    "mag": ("Magahi",  "पानी हमनी के जीवन हे"),
    "hin": ("Hindi",   "पानी हमारा जीवन है"),
}

for code, (name, text) in samples.items():
    model = VitsModel.from_pretrained(f"facebook/mms-tts-{code}")
    tok   = AutoTokenizer.from_pretrained(f"facebook/mms-tts-{code}")
    ids   = tok(text, return_tensors="pt")
    print(f"{name}: {ids['input_ids'].shape[1]} tokens")
    if ids["input_ids"].shape[1] == 0:
        print(f"  ⚠️ empty — skipping"); continue
    with torch.no_grad():
        wav = model(**ids).waveform
    scipy.io.wavfile.write(f"{code}_{name}.wav", rate=model.config.sampling_rate,
                           data=wav.float().numpy().T)
    print(f"  ✅ saved {code}_{name}.wav")
```

```python
!pip install -q transformers torch
!pip install -q git+https://github.com/VarunGumma/IndicTransToolkit.git

import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
from IndicTransToolkit.processor import IndicProcessor

ckpt = "ai4bharat/indictrans2-indic-indic-dist-320M"
tok   = AutoTokenizer.from_pretrained(ckpt, trust_remote_code=True)
model = AutoModelForSeq2SeqLM.from_pretrained(ckpt, trust_remote_code=True)
ip    = IndicProcessor(inference=True)

src, tgt = "hin_Deva", "sat_Olck"          # Hindi -> Santali (Ol Chiki)

sents = [
    "किसान खेत में गेहूँ उगाता है और उसे बाज़ार में बेचता है।",   # original textbook line
    "किसान खेत में धान उगाता है।",                              # our adapted, sentence 1
    "धान हाट में बिकता है।",                                    # sentence 2
    "किसान पैसे कमाता है।",                                     # sentence 3
]

batch = ip.preprocess_batch(sents, src_lang=src, tgt_lang=tgt)
enc   = tok(batch, truncation=True, padding="longest", return_tensors="pt")
with torch.no_grad():
    out = model.generate(**enc, max_length=256, num_beams=5)
dec = tok.batch_decode(out, skip_special_tokens=True)
for s, t in zip(sents, ip.postprocess_batch(dec, lang=tgt)):
    print(s, "\n  →", t, "\n")
```
